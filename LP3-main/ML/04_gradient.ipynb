```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# define the function and its derivative

def f(x) :
    return (x+3)**2

def derivative(x) :
    return 2 * (x+3)
```

```python
# implement gradient descent

x = 2
learning_rate = 0.1
iterations = 100

for i in range(iterations) :
    x = x - learning_rate * derivative(x)

print("Local Minimum x : ", x)
print("Minimum value y : ", f(x))
```

```python
# plot the function

x_values = np.linspace(-6, 3, 100)
y_values = f(x_values)

plt.plot(x_values, y_values)
plt.scatter(x, f(x), color='red', label='Minimum')
plt.xlabel('x')
plt.ylabel('y')
plt.title('Gradient Descent')
plt.legend()
plt.grid()
plt.show()
```
