```python
import pandas as pd
import numpy as np
```

```python
# Load Dataset
```

```python
df = pd.read_csv('sales_data_sample.csv', encoding='latin1')
df.head()
```

```python
# 3 Preprocess the dataset
```

```python
x = df[['QUANTITYORDERED', 'SALES']]
x = x.dropna()
```

```python
# normalization of data
```

```python
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()

x_scaled = scaler.fit_transform(x)
```

```python
# Find optimal clusters using Elbow Method
```

```python
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans, AgglomerativeClustering

wcss = []

for i in range(1, 11) :
    kmeans = KMeans(n_clusters=i, random_state=42, n_init=10)
    kmeans.fit(x_scaled)
    wcss.append(kmeans.inertia_)

plt.plot(range(1,11), wcss, marker='o')

plt.title('Elbow Method')
plt.xlabel('Number of Clusters')
plt.ylabel('WCSS')
plt.show()
```

```python
# implement K-Mean clustering
```

```python
kmeans = KMeans(n_clusters=3, random_state=42, n_init=10)

y_kmeans = kmeans.fit_predict(x_scaled)

print("K-Means Cluster Labels:")
print(y_kmeans)
```

```python
# visualize cluster
plt.scatter(
    x_scaled[:,0], x_scaled[:,1], c = y_kmeans, cmap='rainbow'
)

plt.title('K-Means Clustering')
plt.xlabel('Quantity Ordered (Scaled)')
plt.ylabel('Sales (Scaled)')
plt.show()
```

```python
# Implement Hierarchical Clustering
```

```python
hc = AgglomerativeClustering(n_clusters=3, linkage='ward')
y_hc = hc.fit_predict(x_scaled)

print("Hierarchical Clustering:")
print(y_hc)
```

```python
# Display of cluster visualization
plt.scatter(
    x_scaled[:,0],
    x_scaled[:,1],
    c=y_hc,
    cmap='rainbow'
)

plt.title('Hierarchical Clustering')
plt.xlabel('Quantity Ordered (Scaled)')
plt.ylabel('Scales (Scaled)')
plt.show()
```

```python
print("K-Means Cluster Counts:")
print(pd.Series(y_kmeans).value_counts())

print("Hierarchical Cluster Counts:")
print(pd.Series(y_hc).value_counts())
```
